```python
# metin temizleme fonksiyon kullanımı
def metin_temizleme(metin):

  metin = metin.lower() # Küçük harfe dönüştürme

  metin = re.sub(r"http\S+|www\S+", "", metin) # URL'leri kaldırma

  metin = BeautifulSoup(metin, "html.parser").get_text() # HTML etiketlerini temizleme

  metin=re.sub(r"[^a-zA-ZçğıöşüôÇĞİÖŞÜ0-9\s]","",metin) # Türkçe karakterleri koruyarak özel karakterleri kaldırma

  metin = re.sub(r'\s+', ' ', metin).strip() # Fazla boşlukları tek boşluğa indirme
  return metin

# 'metin' değişkeninin string olduğundan emin olalım
metin_string = str(metin)
print("temizlenmeden önceki hali :  ", metin)
temizlenmis_metin = metin_temizleme(metin_string)
print("temizlenmiş metin :  ", temizlenmis_metin)
```

```python
#3.hafta konuları
#stemming kök bulma
import nltk
nltk.download("wordnet") # nltk nın wordnet sözlük veri tabanını indirir.
from nltk.stem import PorterStemmer
words=["running","runs","ran","runs","better","go","went","begining"]
stemmer=PorterStemmer()#stemmer için nesnelei oluşturur.
stems=[stemmer.stem(word) for word in words]
print(stems)
```

```python
#Lemmatizer dilbilgisine göre kök bulma
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
words=["running","runs","ran","runs","better","go","went","begining"]
lemmas=[lemmatizer.lemmatize(word, pos="v") for word in words]
print(lemmas)
```

```python
# BOW  veri seti uygulaması
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
import re
from collections import Counter

if 'file_content' in locals() and file_content is not None:
    # Assuming each line is a document
    documents = file_content.strip().split('\n')
    df = pd.DataFrame({'text': documents})
    display(df.head())

    def clean_text(text):
        text = str(text) # Ensure text is a string
        text = text.lower() #büyük küçük harf çevirme
        text = re.sub(r"\d+","",text)#rakamları temizleme
        text = re.sub(r"[^\w\s]","",text)#özel karakterlerin kaldırılması
        text = " ".join([word for word in text.split() if len(word)>2])#kısa kelimelerin kaldırılması
        return text

    clean_doc = [clean_text(row) for row in df['text']] #metinleri temizle

    vectorize = CountVectorizer()# verktörel tanımlama
    x = vectorize.fit_transform(clean_doc) # metin sayısal hale gelir
    feature_names = vectorize.get_feature_names_out()# kelime kümesi oluşturma
    print("kelime kümesi= ",feature_names)
    vektor_temsili = x.toarray()#vektor temsili
    print("vektor temsili= ",vektor_temsili)
    df_bow=pd.DataFrame(vektor_temsili,columns=feature_names)
    display(df_bow)
else:
    print("File content not loaded. Please run the file upload and decoding cells first.")
```
